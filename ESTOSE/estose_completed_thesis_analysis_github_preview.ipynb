## DEA Model

            Each farmer was treated as one decision-making unit (DMU). The input-oriented DEA model used:

            | Role | Variable |
            |---|---|
            | Input | Land area |
            | Input | Number of bearing trees |
            | Input | Organic fertilizer |
            | Input | Inorganic fertilizer |
            | Input | Chemicals |
            | Input | Labor |
            | Output | Jackfruit output in kg/tree/year |

            CRS technical efficiency represents overall technical efficiency, VRS technical efficiency represents pure technical efficiency, and scale efficiency is computed as `CRS_TE / VRS_TE`.


```python
dea_summary = pd.read_csv(OUT / "dea_summary.csv")
            rts = pd.read_csv(OUT / "returns_to_scale_count.csv")
            dea_summary, rts
```

## Regression Model

            The regression stage used **VRS technical efficiency** as the dependent variable. Scale efficiency was not used as the dependent variable because the research objective concerned factors associated with technical efficiency, not scale performance.

            Primary predictors:

            - Age
            - Education years
            - Log monthly income
            - Years of jackfruit farming
            - Farm area

            The primary regression was a two-limit Tobit-style censored model, with OLS retained only as a diagnostic robustness check.


```python
tobit = pd.read_csv(OUT / "tobit_regression_results.csv")
            ols = pd.read_csv(OUT / "ols_robustness_results.csv")
            tobit, ols
```

## Regression Diagnostics
